## Read the total profit of all months and show it using the Bar plot. 


```python
# Assigning x and y coordinates
# If plotting directly from df:
# If plotting directly from df:
plt.bar(df['month_number'], df['total_profit'], color='blue', linewidth=3)
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.title('Company Profit per Month')

plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## Read the total profit of all of the months, and show the line plot with the following style properties:

```python
## Extract data from data frame
x = df['month_number']
y = df['total_profit']

## Plot with specified style 
plt.plot(
  x,
  y,
  label='Profit data of last year',
  color= 'red',
  linestyle= 'dotted',
  marker='o',
  markerfacecolor='black',
  markeredgecolor='black',
  linewidth=3
)

# Axis labels and title
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')

# Legend ae the lower_right location
plt.legend(loc='lower right')
           
```

## Print all of the product sales data and show it using a multi-line plot

```python
import matplotlib.pyplot as plt

# Plot all 6 product columns
plt.plot(df['month_number'], df['facecream'], label='Face cream Sales Data', marker='o', linewidth=3)
plt.plot(df['month_number'], df['facewash'], label='Face Wash Sales Data', marker='o', linewidth=3)
plt.plot(df['month_number'], df['toothpaste'], label='Toothpaste Sales Data', marker='o', linewidth=3)
plt.plot(df['month_number'], df['bathingsoap'], label='Bathingsoap Sales Data', marker='o', linewidth=3)
plt.plot(df['month_number'], df['shampoo'], label='Shampoo Sales Data', marker='o', linewidth=3)
plt.plot(df['month_number'], df['moisturizer'], label='Moisturizer Sales Data', marker='o', linewidth=3)

# Axis labels and title
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.title('Sales Data')

# Set ticks and ranges
plt.xticks(df['month_number'])
plt.yticks([1000, 2000, 4000, 6000, 8000, 10000, 12000, 15000, 18000])

# Legend at upper-left
plt.legend(loc='upper left')

plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## Read “bathingsoap” sales data for each month and show it using a scatter plot.


```python
# 1. Create the scatter plot
plt.scatter(df['month_number'], df['bathingsoap'])

# 2. Labels and Title
plt.xlabel('Month Number')
plt.ylabel('Number of Units Sold')
plt.title('Bathing Soap Sales Data')

# 3. Format X-ticks to show all months (1 through 12)
plt.xticks(df['month_number'])

# 4. Add the grid with the required line style "--"
plt.grid(True, linestyle="--", linewidth=1)

# 5. Display the plot
plt.show()
```

## Given Data Set
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]

Show the average weight against the average height (use a line chart or any other chart):
The generated line plot must include the following Style properties:
The line style should be dash-dot and the line color should be green.
Show the legend at the lower-right location.
X label name = Weight in kg.
Y label name = Height in cm.
Title = Average weight with respect to average height.
Add a circle marker: marker size = 10; circle marker color = green.


```python
import matplotlib.pyplot as plt

height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]

# Plot weight (x-axis) vs height (y-axis) with the requested styling
plt.plot(
    weight,
    height,
    label='Height vs Weight',
    color='green',
    linestyle='dashdot',       # dash-dot line style (or '-.')
    marker='o',                # circle marker
    markersize=10,             # marker size = 10
    markerfacecolor='green',   # circle marker color = green
    markeredgecolor='green'
)

# Axis labels and Title
plt.xlabel('Weight in kg.')
plt.ylabel('Height in cm.')
plt.title('Average weight with respect to average height.')

# Legend at the lower-right location
plt.legend(loc='lower right')

# Display the plot
plt.show()
```
